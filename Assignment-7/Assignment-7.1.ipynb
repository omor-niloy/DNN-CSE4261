```python
from ultralytics import YOLO
import cv2
import time
source = "/home/cse/Documents/NP/promod/4-2/Assignment-7/new_york.mp4"
```

```python
model = YOLO("yolov8n.pt")
results = model(source, save = True, name = 'v8')
```

```python
model = YOLO("yolo11n.pt")
results = model(source, save = True, name = "v11")
```

```python
model = YOLO("yolo12n.pt")
results = model(source, save = True, name = "v12")
```

```python
def model_evaluate(model_name):
    model = YOLO(model_name)
    vfile = cv2.VideoCapture(source)
    total_object_count = 0
    start = time.time()
    while vfile.isOpened():
        ret, frame = vfile.read()
        if not ret:
            break
        results = model(frame, verbose=False)
        for result in results:
            total_object_count += len(result.boxes)

    print(f"Time taken to process the video: {time.time() - start:.2f} seconds")
    vfile.release()
    print(f"Total number of object instances detected by {model_name} across the video: {total_object_count}\n")
```

```python
model_names = ['yolov8n.pt', 'yolo11n.pt', 'yolo12n.pt']
for model_name in model_names:
    model_evaluate(model_name)
```
